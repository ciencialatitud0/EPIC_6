## 1. ¿Por qué medir la rotación de una galaxia?

Una estrella que orbita a distancia $r$ del centro se mueve con velocidad

$$V(r) = \sqrt{\frac{G\,M(<r)}{r}},$$

donde $M(<r)$ es **toda** la masa que hay dentro de su órbita, brille o no.
Medir la velocidad a cada radio es como *pesar* la galaxia por capas.

- Si casi toda la masa estuviera donde está la luz, lejos del centro la velocidad debería **bajar** (como los planetas lejanos del Sol, que van más despacio).
- Si la velocidad se mantiene **alta y casi constante**, hay masa que no vemos.

## 5. ¿Cuánta materia oscura hace falta?

La gravedad que falta se atribuye a la **materia oscura**:

$$V_{DM} = \sqrt{V_{obs}^2 - V_{bar}^2}
\qquad\qquad
f_{DM} = 1 - \left(\frac{V_{bar}}{V_{obs}}\right)^2$$
