## 3. Estimen tres centros de tránsito

### Pregunta científica

¿Con qué frecuencia se repiten las caídas de brillo observadas en WASP-18?

Para estimar el período orbital analizaremos tres tránsitos consecutivos. Las ventanas de tiempo serán proporcionadas para que el equipo se concentre en medir e interpretar el patrón.

En cada ventana:

1. Seleccionaremos las mediciones aceptadas.
2. Identificaremos los diez puntos con menor flujo.
3. Utilizaremos la mediana de sus tiempos para estimar el centro del tránsito.

Si los centros estimados son $t_1$, $t_2$ y $t_3$, calcularemos:

$$
P_1 = t_2 - t_1
$$

$$
P_2 = t_3 - t_2
$$

Finalmente, estimaremos el período orbital mediante:

$$
P = \frac{P_1 + P_2}{2}
$$

## 4. Calculen la fase orbital

### Pregunta científica

¿Cómo podemos comparar observaciones obtenidas durante muchas órbitas diferentes?

La curva de luz completa contiene varios tránsitos del mismo planeta. El **plegado por fase** coloca juntas las mediciones que corresponden al mismo momento de cada órbita.

Para hacerlo utilizaremos:

- el período orbital estimado, $P$;
- uno de los centros de tránsito como tiempo de referencia, $t_0$;
- el tiempo de cada observación, $t$.

La fase orbital se calculará mediante:

$$
\phi =
\frac{\left[(t-t_0+P/2)\bmod P\right]}{P}
-\frac{1}{2}
$$

Con esta definición:

- la fase $\phi=0$ corresponde aproximadamente al centro del tránsito;
- las fases negativas representan momentos anteriores al centro;
- las fases positivas representan momentos posteriores;
- observaciones separadas por una órbita completa quedan cerca de la misma fase.

No es necesario deducir esta expresión. El objetivo es aplicarla en Python y explicar qué consigue.

### Checkpoint

Todos los integrantes deben poder explicar por qué el plegado permite superponer observaciones tomadas durante órbitas diferentes.

```python
# Usar el primer centro de tránsito como referencia
t0 = t1

# Calcular la fase orbital de cada medición aceptada
phase = (
    (
        accepted_data["time_days"]
        - t0
        + 0.5 * period
    )
    % period
) / period - 0.5

print(f"Tiempo de referencia, t0: {t0:.5f} días")
print(f"Fase mínima: {phase.min():.5f}")
print(f"Fase máxima: {phase.max():.5f}")

# Mostrar algunos valores calculados
pd.DataFrame(
    {
        "time_days": accepted_data["time_days"],
        "phase": phase,
        "relative_flux": accepted_data["relative_flux"],
    }
).head()
```

```python
# Curva de luz plegada alrededor del tránsito
fig, ax = plt.subplots(figsize=(9, 5))

ax.scatter(
    phase,
    accepted_data["relative_flux"],
    s=7,
    alpha=0.35,
    color="tab:blue",
)

# Marcar el centro de referencia
ax.axvline(
    0,
    color="tab:red",
    linestyle="--",
    linewidth=1,
    label="Centro de referencia",
)

# Mostrar únicamente la región cercana al tránsito
ax.set_xlim(-0.12, 0.12)

ax.set_xlabel("Fase orbital")
ax.set_ylabel("Flujo relativo")
ax.set_title("Curva de luz plegada de WASP-18 b")
ax.legend()
ax.grid(alpha=0.2)

plt.tight_layout()
plt.show()
```

## 5. ¿Cuánta luz desaparece?

### Pregunta científica

¿Qué fracción de la luz de WASP-18 queda bloqueada durante el tránsito?

Compararemos el flujo relativo dentro del tránsito con el flujo observado cuando el planeta no está pasando frente a la estrella.

Utilizaremos estas regiones de fase:

- **Dentro del tránsito:** desde $\phi=-0.03$ hasta $\phi=0.03$.
- **Fuera del tránsito:** desde $\phi=-0.12$ hasta $\phi=-0.08$ y desde $\phi=0.08$ hasta $\phi=0.12$.

Calcularemos dos medianas:

- $F_{\mathrm{in}}$: mediana del flujo dentro del tránsito.
- $F_{\mathrm{out}}$: mediana del flujo fuera del tránsito.

La profundidad del tránsito se estimará mediante:

$$
\delta =
\frac{F_{\mathrm{out}}-F_{\mathrm{in}}}
{F_{\mathrm{out}}}
$$

También expresaremos el resultado como porcentaje:

$$
\delta_{\%}=100\delta
$$

Utilizamos la mediana porque es menos sensible a mediciones individuales inusuales que el promedio.

### Checkpoint

Todos los integrantes deben poder explicar por qué el flujo dentro del tránsito es menor que el flujo fuera del tránsito.

```python
# Región dentro del tránsito
in_transit_mask = (
    (phase >= -0.03)
    & (phase <= 0.03)
)

# Regiones fuera del tránsito, a ambos lados
out_of_transit_mask = (
    (
        (phase >= -0.12)
        & (phase <= -0.08)
    )
    |
    (
        (phase >= 0.08)
        & (phase <= 0.12)
    )
)

print(f"Mediciones dentro del tránsito: {in_transit_mask.sum()}")
print(f"Mediciones fuera del tránsito: {out_of_transit_mask.sum()}")
```

```python
# Mediana del flujo dentro del tránsito
F_in = np.median(
    accepted_data.loc[in_transit_mask, "relative_flux"]
)

# Mediana del flujo fuera del tránsito
F_out = np.median(
    accepted_data.loc[out_of_transit_mask, "relative_flux"]
)

# TAREA 5: calcula la profundidad como fracción y como porcentaje.
depth = None
depth_percent = None

if depth is None or depth_percent is None:
    raise ValueError(
        "Completa depth y depth_percent con F_in y F_out."
    )

print(f"Flujo mediano dentro del tránsito, F_in: {F_in:.6f}")
print(f"Flujo mediano fuera del tránsito, F_out: {F_out:.6f}")
print(f"Profundidad como fracción: {depth:.6f}")
print(f"Profundidad como porcentaje: {depth_percent:.4f} %")
```

## 6. ¿Qué nos dice esto sobre el planeta?

### Pregunta científica

¿Cómo permite la profundidad del tránsito estimar el tamaño del planeta en comparación con su estrella?

Durante el tránsito, el planeta bloquea una parte del área visible de la estrella. En una aproximación sencilla, la profundidad se relaciona con los radios mediante:

$$
\delta \approx
\left(
\frac{R_{\mathrm{p}}}{R_{\star}}
\right)^2
$$

Por lo tanto, el cociente entre el radio del planeta y el radio de la estrella puede estimarse como:

$$
\frac{R_{\mathrm{p}}}{R_{\star}}
\approx
\sqrt{\delta}
$$

La raíz cuadrada aparece porque la cantidad de luz bloqueada está relacionada con un **área**, mientras que el radio representa una **longitud**.

Esta relación es aproximada. No considera efectos como el oscurecimiento hacia el borde de la estrella ni la trayectoria exacta del planeta durante el tránsito.
