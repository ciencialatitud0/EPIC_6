````python
def graficar_espectros_desplazados(
    longitud_onda,
    flujos_sel,
    temperaturas_sel,
    desplazamiento=0.5,
):
    """
    Grafica varios espectros separados por desplazamientos verticales.

    Parámetros
    ----------
    longitud_onda : array de una dimensión
        Longitudes de onda compartidas por todos los espectros.

    flujos_sel : array de dos dimensiones
        Espectros seleccionados. Cada fila contiene un espectro.

    temperaturas_sel : array de una dimensión
        Temperatura correspondiente a cada espectro seleccionado.

    desplazamiento : float
        Separación vertical entre espectros consecutivos.
    """

    fig, ax = plt.subplots(figsize=(10, 7))

    for i, flujo in enumerate(flujos_sel):
        flujo_desplazado = flujo + i * desplazamiento

        ax.plot(
            longitud_onda,
            flujo_desplazado,
            label=f"{temperaturas_sel[i]:.0f} K",
        )

    ax.set_xlabel(r"Longitud de onda ($\mathrm{\AA}$)")
    ax.set_ylabel("Flujo escalado + desplazamiento vertical")

    return fig, ax
````

````python
fig, ax = graficar_espectros_desplazados(long_onda, flujos_sel, temperaturas_sel, desplazamiento=0.6,)

ax.axvline(hbeta, color='black', linestyle="--", label=r"H$\beta$: $4861.3$ (A)")

ax.set_xlim(4780, 4940)
ax.legend()

plt.show()
````

### Calcular el área

El espectro contiene valores de flujo en diferentes longitudes de onda.

`np.trapezoid` estima el área al unir flujos circundantes con líneas rectas y sumar las áreas trapezoidales resultantes. Es una forma de integrar el área de interés.

Para calcular

```python
np.trapezoid(y, x)
```

donde `y` es el déficit de flujo y `x` es la longitud de onda.

POr lo tanto, estamos integrando el déficit de flujo sobre el rango de longitud de onda correspondiente a la ventana de medición.

````python
intensidad_linea = np.trapezoid(deficit_linea, long_onda_linea)

print("Amplitud de H-beta pseudo-equivalente:", intensidad_linea, " Å")
````
