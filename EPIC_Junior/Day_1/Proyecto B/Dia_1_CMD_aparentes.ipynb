El color observacional será:

$$G_{\rm BP}-G_{\rm RP}.$$

Un valor menor corresponde generalmente a una estrella más azul y caliente; un valor mayor, a una estrella más roja y fría. La escala de magnitudes funciona al revés: **una magnitud menor significa mayor brillo**. Por eso invertiremos el eje vertical.

```python
# TAREA 2: calculen BP − RP.
stars["bp_minus_rp"] = None

if stars["bp_minus_rp"].isna().all():
    raise ValueError("Calculen bp_minus_rp antes de continuar.")

```

```python
# TAREA 3: conserven únicamente las filas con use_star=True.
selected = None

if not isinstance(selected, pd.DataFrame) or selected.empty:
    raise ValueError("Creen el DataFrame selected y revisen la condición.")

print(f"Color calculado y {len(selected)} estrellas preparadas.")
```

```python
# Recuperar las estrellas seleccionadas
selected = stars.loc[stars["use_star"]].copy()

fig, axes = plt.subplots(1, 2, figsize=(13, 5.5), sharex=True, sharey=True)

for ax, cluster in zip(axes, CLUSTERS):

    group = selected.loc[selected["cluster"] == cluster]

    ax.scatter(group["bp_minus_rp"], group["g_mag"], s=24, alpha=0.72, color=COLORS[cluster], edgecolor="none")

    ax.set_title(LABELS[cluster])
    ax.set_xlabel(r"Color observado $G_{BP}-G_{RP}$ (mag)")

axes[0].set_ylabel("Magnitud aparente G (mag)")

# Escalas iguales y magnitudes brillantes arriba
axes[0].set_xlim(-0.5, 3.6)
axes[0].set_ylim(19, 4.5)

fig.suptitle("Figura 1. CMD aparentes", fontsize=16, fontweight="bold")
fig.tight_layout()
fig.savefig("figure_1_apparent_cmds.png", dpi=220, bbox_inches="tight")
plt.show()
print("Figura guardada como figure_1_apparent_cmds.png")
```
