## Tema 1: Python como calculadora — el péndulo

Recuerda la fórmula del período de un péndulo simple:

$$T = 2\pi \sqrt{\frac{L}{g}}$$

Usemos Python para calcularla en vez de hacerlo a mano.

```python
import math

L = 1.0    # longitud del péndulo, en metros
g = 9.81   # gravedad en la Tierra, en m/s^2

T = 2 * math.pi * math.sqrt(L / g)
print(f"Para L = {L} m en la Tierra, el período es T = {T:.2f} s")
```

## Tema 2: Python para leer datos — movimiento parabólico (30°)

Ahora vamos a trabajar con datos reales de un experimento.

1. Abre el simulador de movimiento parabólico: 🔗 **https://cphysplus.github.io/fisica-interactiva/movimiento-parabolico.html**
2. Ajusta el ángulo de lanzamiento a **30°**.
3. Presiona reproducir el lanzamiento y luego **"Exportar Trayectoria (CSV)"**. Esto descargará un archivo a tu computadora.
4. Ejecuta la celda de abajo y súbelo cuando aparezca el botón **"Elegir archivos"**.

```python
import pandas as pd
from google.colab import files

print("Selecciona el archivo CSV del lanzamiento a 30°:")
subido_30 = files.upload()
```

```python
nombre_archivo_30 = list(subido_30.keys())[0]
datos_30 = pd.read_csv(nombre_archivo_30)

print(f"Se cargaron {len(datos_30)} filas.")
datos_30.head()
```

La tabla tiene columnas: el **tiempo** (`t`), la **posición horizontal** (`x`) y **vertical** (`y`), y las **velocidades** horizontal (`vx`) y vertical (`vy`) del proyectil en cada instante. Para graficar la trayectoria usaremos `x` y `y`.

## Tema 3: Python para graficar — la trayectoria del proyectil (30°)

Grafiquemos la posición Y contra la posición X para ver la forma de la trayectoria (la parábola).

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.plot(datos_30["x"], datos_30["y"], color="#6c5ce7", linewidth=2, label="Ángulo = 30°")
plt.xlabel("Posición X (m)")
plt.ylabel("Posición Y (m)")
plt.title("Trayectoria del proyectil — Ángulo de 30°")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

### Grafica y compara ambos ángulos

Grafica las dos trayectorias (30° y 75°) en el mismo gráfico para compararlas.

```python
plt.figure(figsize=(8, 5))
plt.plot(datos_30["x"], datos_30["y"], color="#6c5ce7", linewidth=2, label="Ángulo = 30°")
plt.plot(datos_75["x"], datos_75["y"], color="#e74c3c", linewidth=2, label="Ángulo = 75°")
plt.xlabel("Posición X (m)")
plt.ylabel("Posición Y (m)")
plt.title("Comparación de trayectorias: 30° vs. 75°")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```
