# Tema 1: Python como calculadora 

## 1a: Cálculemos el tamaño de un agujero negro

Un agujero negro no tiene superficie, pero sí tiene una frontera: el **horizonte de eventos**.

Todo lo que cruza esa frontera ya no puede salir, ni siquiera la luz.

Su radio se calcula con una fórmula muy sencilla:

$$R = 2.95\ \text{km} \times M$$

donde $M$ es la masa del agujero negro medida **en masas Solares** (es decir, cuántos "Soles" de materia tiene).

```python
masa = 10        # masa del agujero negro, en masas Solares

radio = 2.95 * masa    # radio del horizonte de eventos, en km

print(f"Un agujero negro de {masa} masas solares tiene un radio de {radio} km.")
```

## 1b: ¿Qué tan densa es una estrella de neutrones?

Una estrella de neutrones tiene más masa que el Sol, pero mide apenas unos **12 km de radio**: cabría entre Ambato y Pelileo.

La densidad es simplemente la masa dividida para el volumen:

$$\text{densidad} = \frac{\text{masa}}{\text{volumen}}, \qquad
\text{volumen} = \frac{4}{3}\pi R^3$$

```python
masa_kg = 1.4 * 2e30      # 1.4 masas solares, en kilogramos
radio_m = 12 * 1000       # 12 km pasados a metros

# Formula del volumen
volumen = (4/3) * 3.1416 * radio_m**3
densidad = masa_kg / volumen

print(f"Volumen:  {volumen:.2e} metros cúbicos")
print(f"Densidad: {densidad:.2e} kg por metro cúbico")
print(f"Es {densidad/1000:.2e} veces más densa que el agua")
```

## Tema 3: Python para graficar — el diagrama H–R

Hace más de 100 años, dos astrónomos (Hertzsprung y Russell) pusieron las estrellas en un gráfico de **temperatura** contra **luminosidad**.


Descubrieron que las estrellas no se reparten al azar: forman **grupos**.

Dos detalles del gráfico:

- Usamos **escala logarítmica** porque las diferencias entre los números son gigantes (de 0.0001 a 200000 veces la luz del Sol).

- El eje de temperatura va **al revés**: las estrellas calientes a la izquierda.

Así lo dibujaron ellos, y así se sigue usando hoy.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))
plt.scatter(estrellas["temperatura"], estrellas["luminosidad"], s=70, color="#6c5ce7")

plt.xscale("log")
plt.yscale("log")
plt.gca().invert_xaxis()          # las calientes van a la izquierda

plt.xlabel("Temperatura (K)")
plt.ylabel("Luminosidad (Sol = 1)")
plt.title("Diagrama H-R")
plt.grid(alpha=0.3)
plt.show()
```

#### Grupos de estrellas:

¿Ves los grupos? Pintemos cada tipo de estrella de un color distinto para que queden más claros.

```python
plt.figure(figsize=(9, 6))

for nombre_grupo in estrellas["grupo"].unique():
    g = estrellas[estrellas["grupo"] == nombre_grupo]
    plt.scatter(g["temperatura"], g["luminosidad"], s=70, label=nombre_grupo)

plt.xscale("log")
plt.yscale("log")
plt.gca().invert_xaxis()

plt.xlabel("Temperatura (K)")
plt.ylabel("Luminosidad (Sol = 1)")
plt.title("Diagrama H-R por grupo de estrellas")
plt.legend(fontsize=8)
plt.grid(alpha=0.3)
plt.show()
```

### Bonus: comprobemos el radio con una fórmula

La respuesta a las dos preguntas anteriores es el **tamaño**. Y lo bonito es que
el radio se puede **calcular** con una sola línea de Python, a partir de la
temperatura y la luminosidad:

$$R = \left(\frac{5772}{T}\right)^2 \times \sqrt{L}$$

(el resultado sale en radios Solares: Sol = 1)

Como el simulador ya nos dio el radio medido, podemos comparar las dos columnas y ver si la fórmula funciona.

```python
estrellas["radio_calculado"] = (5772 / estrellas["temperatura"])**2 * estrellas["luminosidad"]**0.5

estrellas[["nombre", "grupo", "temperatura", "luminosidad", "radio", "radio_calculado"]]
```
